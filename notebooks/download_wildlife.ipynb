# FAA Wildlife Strikes Download
Use this notebook to pull the FAA wildlife-strikes dataset into the local `data/` directory without storing notebooks alongside the raw files.

```python
from pathlib import Path
DATA_DIR = Path.cwd() / "data"
DATA_DIR.mkdir(exist_ok=True)
DATA_DIR
```

## Download Dataset with kagglehub
This step pulls the FAA dataset into the cache, then copies it under `data/` so source control can track it if needed.

```python
import os
import shutil
import kagglehub
os.environ.setdefault("KAGGLE_CONFIG_DIR", str(Path.cwd()))
cache_path = kagglehub.dataset_download("faa/wildlife-strikes")
dataset_dir = DATA_DIR / "wildlife-strikes"
shutil.copytree(cache_path, dataset_dir, dirs_exist_ok=True)
dataset_dir
```

## Inspect Downloaded Files
List a few of the files to confirm the copy worked and only raw data lives under `data/`.

```python
from itertools import islice
file_listing = sorted(p.relative_to(DATA_DIR) for p in dataset_dir.rglob("*") if p.is_file())
list(islice(file_listing, 10))
```
